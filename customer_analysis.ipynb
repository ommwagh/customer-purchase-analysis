# 🛍️ Customer Purchase Behavior Analysis

This project analyzes customer purchase behavior from a retail dataset using Python and pandas. It calculates key metrics such as total revenue, customer count, average order value, and high-value customer identification.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('customer_data.csv')
df.head()
```

## 🔍 Basic Summary Stats

```
# Total revenue
total_revenue = df['Order_Amount'].sum()
print(f'Total Revenue: £{total_revenue}')

# Number of unique customers
unique_customers = df['Customer_ID'].nunique()
print(f'Number of Unique Customers: {unique_customers}')

# Average order value
avg_order_value = df['Order_Amount'].mean()
print(f'Average Order Value: £{avg_order_value:.2f}')
```

## 💎 High-Value Customers

```
# Define threshold for high-value customers
threshold = 200

# Group by Customer and sum their order amounts
customer_summary = df.groupby('Customer_ID')['Order_Amount'].sum().reset_index()

# Filter high-value customers
high_value_customers = customer_summary[customer_summary['Order_Amount'] > threshold]
print(f'Number of High-Value Customers: {len(high_value_customers)}')
high_value_customers.head()
```

## 📊 Revenue by Region

```
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Region', y='Order_Amount', estimator=sum, ci=None)
plt.title('Total Revenue by Region')
plt.ylabel('Revenue (£)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
